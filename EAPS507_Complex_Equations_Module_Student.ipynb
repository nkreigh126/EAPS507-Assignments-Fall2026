# Equation 1 — Potential Temperature

\[
$\theta = T\left(\frac{p_0}{p}\right)^{R_d/c_p}$
\]

where ($\theta$) is potential temperature (K), (T) is temperature (K), (p) is pressure (hPa), ($p_0$=1000\) hPa, \($R_d$=287.05 Jkg$^{-1}$ K$^{-1}$), and ($c_p=1004.0 J kg^{-1} K^{-1}$).


### Task
Write:

```python
potential_temperature(T, p)
```

Test with `T = 288 K` and `p = 850 hPa`.

````python
# import atmos_equations as ae
# Test Equation 1 here

def potential_temperature_v1(T, p, p0=1000, Rd=287.05, Cp=1004.0):
    """
    This function computes the potential temperature for a specific temperature, T.
    Below are the specific definitions of the variables the function takes.

    Arguements:

    T: Temperature in Kelvin (K)
    P:Pressure in hectopascals (hPa)
    Rd: Dry air ideal gas constant (J/kg/K)
    Cp: Specific heat capacity at constant pressure (J/kg/K)

    Return
    _______

    This function returns the potential temperature give sa ${theta}$

    """
    expo = Rd / Cp  #exponent of equation
    print(expo)
    multiplication_part = (p0/p) #main part of equation
    
    theta = T* multiplication_part **expo #combine parts of equation
    
    return theta #return value of theta
    
````

# Equation 2 — Saturation Vapor Pressure

Use the Bolton approximation:

\[
$e_s(T_c)=6.112\exp\left(\frac{17.67T_c}{T_c+243.5}\right)$
\]

where \(e_s\) is saturation vapor pressure (hPa) and \($T_c$\) is temperature in °C.

### Task
Write:

```python
saturation_vapor_pressure(Tc)
```

Test with `Tc = 20 °C`.

````python
import math

def saturation_vapor_pressure_v1(Tc):
    """
    Computes the saturation vapor pressure over liquid water using Bolton's (1980) approximation.

    Parameters:
    -----------
    Tc : float, int, or pint.Quantity
        Temperature in degrees Celsius (°C).

    Returns:
    --------
    e_s : float
        Saturation vapor pressure in hectopascals (hPa).
    """
    # Extract magnitude if a MetPy/Pint Quantity is passed
    if hasattr(Tc, 'magnitude'):
        Tc = Tc.magnitude

    # Bolton's approximation formula
    e_s = 6.112 * math.exp((17.67 * Tc) / (Tc + 243.5))
    
    return e_s

# Test the function with Tc = 20 °C
es_20 = saturation_vapor_pressure(20)
print(f"Saturation vapor pressure at 20°C: {es_20:.4f} hPa")
    


````

# Equation 3 — Saturation Mixing Ratio

\[
$w_s=\epsilon\frac{e_s}{p-e_s}$
\]

where \($w_s$\) is saturation mixing ratio ($kg kg^{-1}$), ($\epsilon=0.622$), \($e_s$\) is saturation vapor pressure (hPa), and \(p\) is atmospheric pressure (hPa).

### Task

```python
saturation_mixing_ratio(Tc, p)
```

Your function must call `saturation_vapor_pressure()` rather than rewrite that equation.

Test with `Tc = 20 °C` and `p = 1000 hPa`. Also print the result in g kg\(^{-1}\).

````python
# Test Equation 3 here

#define function saturation_mixing_ratiov1 to computre the saturaiton mixing ratio
def saturation_mixing_ratiov1(Tc,p):

    """
    This function computes the Saturation Mixing Ratio for a specific temperature, Tc, in Celsius and
    a pressure, p in hectopascals.
    
    Below are the specific definitions of the variables the function takes.

    Arguements:

    Tc: Temperature in Celsius (C)
    p: Pressure in Hectopascals (hPa)
    e_s: saturation vapor pressure (hPa)
    w_s: Saturation mixing ratio (kg*kg^-1)
    epsilon: constant 0.622
    Return
    _______

    This function returns the saturation vapor pressure given a Temperature in Celsius.

    """
    #constant epsilon 0.622
    epsilon = 0.622
    #call sautration vapor pressure function to calculate the e_s for the equation
    e_s = saturation_vapor_pressure_v1(Tc)
    # the fraction in the formula, e_s / (p-e_s)
    
    frac = e_s / (p - e_s)

    w_s = epsilon * frac

    print(f"Saturation mixing ratio: {w_s:.6f} kg/kg")
    print(f"Saturation mixing ratio: {w_s * 1000:.2f} g/kg")

    return w_s

# Call the function so the code inside runs and prints
ws = saturation_mixing_ratiov1(20, 1000)
````

# Equation 4 — Hypsometric Equation

\[
$z_2-z_1=\frac{R_d\overline{T_v}}{g}\ln\left(\frac{p_1}{p_2}\right)$
\]

where ($z_2-z_1$) is layer thickness (m), ($R_d=287.05 J kg^{-1} K^{-1}$), ($\overline{T_v}$) is layer-mean virtual temperature (K), ($g=9.80665 ms^{-2})$, ($p_1$) is lower-level pressure, and ($p_2$) is upper-level pressure.

### Task
Write:

```python
hypsometric_thickness(Tv, p1, p2)
```

Test for the 1000–500 hPa layer using `Tv = 270 K`, `p1 = 1000 hPa`, and `p2 = 500 hPa`.

````python
# Test Equation 4 here


# Test Equation 4 here

import math

def hypsometric_thickness_v1(Tv,p1,p2):

    """
    This function computes the Hypsometric Equation for a virtual temperature, Tv, in Celsius and
    a pressure, p1 in hectopascals, and p2 in hectopascals
    
    Below are the specific definitions of the variables the function takes.

    Arguements:

    Tv: Virtual Temperature in Kelvin (K)
    p1: Pressure in Hectopascals (hPa)
    p2: Pressure in Hectopascals (hPa)
    z2 = upper level thickness (m)
    z1 = lower level thickness (m)
    g = gravity (ms^-2)
    Rd = dry air constant (Jkg^-1 K-1)
    Return
    _______

    This function returns the Hypsometric Equation given a layer thickness in meters.

    """
   #Define constants for gravity and dry air gas constant 
    
    g= 9.80665
    Rd = 287.05
    
    # Hypsometric equation: (Rd * Tv / g) * ln(p1 / p2)
    thickness = (Rd * Tv / g) * math.log(p1 / p2)
    
    return thickness

# Test for the 1000-500 hPa layer using Tv = 270 K, p1 = 1000 hPa, and p2 = 500 hPa
dz = hypsometric_thickness_v1(270, 1000, 500)
print(f"Layer thickness (1000-500 hPa): {dz:.2f} m")

    
    
````

# Equation 5 — Geostrophic Wind

\[
$u_g=-\frac{g}{f}\frac{\partial Z}{\partial y}$
\]

\[
$v_g=\frac{g}{f}\frac{\partial Z}{\partial x}$
\]

with

\[
$f=2\Omega\sin\phi$
\]

where ($u_g$) and ($v_g$) are geostrophic wind components ($ms^{-1}$), ($g=9.80665 ms^{-2}$), ($Z$) is geopotential height (m), ($f$) is the Coriolis parameter ($s^{-1}$), ($\Omega=7.2921\times10^{-5} s^{-1}$), and ($\phi$) is latitude.

### Task
Write:

```python
geostrophic_wind(dZdx, dZdy, latitude)
```

The function must return both `ug` and `vg`.

Test with `dZdx = 2.0e-5`, `dZdy = -3.0e-5`, and `latitude = 40°`.

````python
# Test Equation 5 here

def geostrophic_wind_v1(dZdx, dZdy, latitude):
    """
    Computes the zonal (ug) and meridional (vg) components of the geostrophic wind.

    Parameters:
    -----------
    dZdx : float
        Change in geopotential height along x-axis (m/m).
    dZdy : float
        Change in geopotential height along y-axis (m/m).
    latitude : float
        Latitude in degrees.

    Returns:
    --------
    ug, vg : tuple of floats
        Zonal and meridional geostrophic wind components in m/s.
    """
    # Physical constants
    g = 9.80665      # Acceleration due to gravity (m s^-2)
    omega = 7.2921e-5 # Angular velocity of Earth (s^-1)

    # Convert latitude to radians and calculate the Coriolis parameter f = 2 * omega * sin(phi)
    phi = math.radians(latitude)
    f = 2 * omega * math.sin(phi)

    # Calculate wind components
    ug = -(g / f) * dZdy
    vg = (g / f) * dZdx

    return ug, vg

# Test with dZdx = 2.0e-5, dZdy = -3.0e-5, and latitude = 40°
ug, vg = geostrophic_wind_v1(2.0e-5, -3.0e-5, 40)

print(f"Zonal Geostrophic Wind (ug): {ug:.2f} m/s")
print(f"Meridional Geostrophic Wind (vg): {vg:.2f} m/s")
````

# Madhava-Leibniz π calculation


This notebook is designed to be a live demonstration of Python fundamentals applied to calculating π using the [Madhava-Leibniz series](https://en.wikipedia.org/wiki/Approximations_of_%CF%80#Middle_Ages)

The approximation is simple:

$$ \pi \approx 4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} \ldots -(-1)^n \frac{1}{2n - 1}  \right) $$

```python
#Intialize pi at 0
pi =0

#intiate number of iterations -- Made a Large N in order to get close to the value, the smaller the N the less accurate to the true value
N= 10000

#initialize an empty list to store in the pi values -- Nothing in the brackets as we have no initial values to put in
pi_list=[]

#create loops from 1 to n, adding terms to pi -- use for loop to indicate that n-number of time depends on the value N we do it down to 1. 
#If we do 0 iterations, pi is already set to equal 0.
# q is the term 
# if statement is for if n = 1, then the term is equal to 1
#else, run the formula, where n>1, afterwards multiple by 4 as seen Madhava-Leibniz series
#indentation required in order for everything to run under the 'for' and 'in range' function
for n in range (1,N):
    if n==1:
        q = 1
    else :
        q = -((-1)**n) / (2*n - 1)
    pi= pi + q
#keeping pi_list in the loop allows for the pi_list to be updated each time the loop is run
    pi_list.append(4*pi)
#This resets the value of pi to whichever term n is desired.
pi = 4 * pi 
print(f"pi = {pi}")

#The solution file showed ' print(f"pi = {pi:2.7f}") ', which rounds to 7 decimal places, and the 2 being the numbers ahead of the decimal


#Indentation is important, as I found that when I accidentally had pi = 4 * pi indented and under the for / range statement, im pretty sure it looped multiplying by 4 
#resulted in pi = inf

#This checks to ensure the beginning of the list makes sense with the first 5 iterations, as expected, n=1 means pi = 4
print(pi_list[:5])
```

```python
""" Plot the convergence of the series. """

# import the plotting library; give it the shorter name 'plt'  --brings in the plotting system from the package
import matplotlib.pyplot as plt

# plot the terms of the series -- takes the pi terms (pi_list in my case) and plots them
plt.plot(pi_list)

# use a log scale on the x-axis --makes x axis logritimic scale
plt.xscale('log')

# plot a horizontal line at the true value of pi 
plt.axhline(math.pi, color='gray', linestyle='--')

# set a useful title
plt.title("Convergence of the Madhava-Leibniz series for pi")

# draw the plot
plt.show()
```
